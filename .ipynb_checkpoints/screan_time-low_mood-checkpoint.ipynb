```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.linear_model import LogisticRegression,LinearRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.metrics import accuracy_score,r2_score,classification_report
```

```python
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.pipeline import Pipeline
```

```python
transformer=ColumnTransformer(transformers=[
    ('trf1',SimpleImputer(),['age','daily_screen_hours',
                             'minutes_to_first_check_after_waking','avg_sleep_hours','anxiety_score_0to27',
                             'physical_activity_days_per_week']),
    ('tnf2',Pipeline([('imputer', SimpleImputer(strategy='most_frequent')),
                      ('encoder', OneHotEncoder(handle_unknown='ignore'))]),['gender']),
    ('trf3',
         Pipeline([('imputer', SimpleImputer(strategy='most_frequent')),
                   ('encoder', OrdinalEncoder())]),['night_time_use','uses_screen_time_limits','attempted_digital_detox',
                                                    'primary_purpose','occupation'])
],remainder='passthrough')
```

```python
transformer
```

```python
X_train=transformer.fit_transform(X_train)
X_test=transformer.transform(X_test)
```

```python
X_train
```

```python
X_test
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import xgboost as xgb
```

```python
scaler=StandardScaler()
X_train_sc=scaler.fit_transform(X_train)
X_test_sc=scaler.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
```

```python
params = {
    'objective':'binary:logistic',
    'max_depth':4,
    'learning_rate':1,
    'n_estimators':100,
    'alpha':20
}
```

```python
classifier=RandomForestRegressor(n_estimators=100, random_state=44)
classifier.fit(X_train_sc,y_train)
y_pred=classifier.predict(X_test_sc)
```

```python
pipe = Pipeline([
    ('preprocessor', transformer),
    ('scaler', StandardScaler()),
    ('classifier',classifier)
])
```
